## Objective

The objective of this notebook is to ingest all raw datasets required for the Mutual Fund Analytics project.

This notebook performs:

- Loading raw datasets
- Dataset inspection
- Shape analysis
- Data type verification
- Missing value analysis
- Duplicate detection
- Initial data quality assessment

The processed data generated in later notebooks is used for performance analytics, dashboard development, and advanced analytics.

```python
import os
import warnings
import pandas as pd
import numpy as np

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 10)
```

```python
RAW_PATH = "../data/raw"

print("Raw Dataset Folder:")
print(RAW_PATH)
```

```python
csv_files = sorted([
    file
    for file in os.listdir(RAW_PATH)
    if file.endswith(".csv")
])

print(f"Total Datasets : {len(csv_files)}")
csv_files
```

```python
datasets = {}

for file in csv_files:

    path = os.path.join(RAW_PATH, file)

    df = pd.read_csv(path)

    datasets[file] = df

    print(f"{file}")
    print(f"Shape : {df.shape}")
```

```python
summary = []

for name, df in datasets.items():

    summary.append({

        "Dataset": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1],

        "Missing Values": df.isnull().sum().sum(),

        "Duplicate Rows": df.duplicated().sum()

    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
for name, df in datasets.items():

    print("="*70)

    print(name)

    print("="*70)

    display(df.dtypes)
```

```python
summary_df.to_csv(
    "../reports/data_quality_report.csv",
    index=False
)

print("Dataset Summary Saved Successfully")
```
